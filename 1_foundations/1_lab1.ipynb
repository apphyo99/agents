```python
# Check the key - if you're not using OpenAI, check whichever key you're using! Ollama doesn't need a key.

import os
openai_api_key = os.getenv('AZURE_API_KAY')

if openai_api_key:
    print(f"OpenAI API Key exists and begins {openai_api_key[:8]}")
else:
    print("OpenAI API Key not set - please head to the troubleshooting guide in the setup folder")
    
```

```python
# And now - the all important import statement
# If you get an import error - head over to troubleshooting in the Setup folder
# Even for other LLM providers like Gemini, you still use this OpenAI import - see Guide 9 for why

from openai import OpenAI
from openai import AzureOpenAI
```

```python
# Create a list of messages in the familiar OpenAI format

messages = [{"role":"system", "content":"You are a helpful assistant"},{"role": "user", "content": "What is 2+2?"}]
deployment_name = "gpt-4o-mini"

client = AzureOpenAI(
    api_version="2024-12-01-preview",
    azure_endpoint=os.getenv("AZURE_API_ENDPOINT"),
    api_key=openai_api_key
)
```

```python
# And now call it! Any problems, head to the troubleshooting guide
# This uses GPT 4.1 nano, the incredibly cheap model
# The APIs guide (guide 9) has exact instructions for using even cheaper or free alternatives to OpenAI
# If you get a NameError, head to the guides folder (guide 6) to learn about NameErrors - always instantly fixable

response = client.chat.completions.create(
    messages = messages,
    max_tokens=4096,
    temperature=1.0,
    top_p=1.0,
    model=deployment_name
)

print(response.choices[0].message.content)
```

```python
# And now - let's ask for a question:

def generate_chat_completion(messages, deployment_name, max_tokens=4096, temperature=1.0, top_p=1.0):
    """
    Generates a chat completion using Azure OpenAI.

    Args:
        messages (list): Conversation messages in OpenAI chat format.
        deployment_name (str): The model deployment name.
        max_tokens (int): Maximum tokens to generate.
        temperature (float): Sampling temperature.
        top_p (float): Nucleus sampling parameter.

    Returns:
        str: The model's reply text.
    """
    response = client.chat.completions.create(
        messages=messages,
        model=deployment_name,
        max_tokens=max_tokens,
        temperature=temperature,
        top_p=top_p
    )
    return response.choices[0].message.content
```

```python
# ask it - this uses GPT 4.1 mini, still cheap but more powerful than nano

question = "Please propose a hard, challenging question to assess someone's IQ. Respond only with the question."
messages = [{"role": "user", "content": question}]

response_question = generate_chat_completion(messages, deployment_name )
print(response_question)
```

```python
# form a new messages list
messages = [{"role": "user", "content": response_question}]
```

```python
# Ask it again

answer = generate_chat_completion(messages, deployment_name )
print(answer)
```
